- **employees**: Emp_ID, Emp_Name, Designation, Region, City, Status

- **customerdetail** (view): client_id, client_name, city, state, product_id, amount

> Hint: 
```sql
WITH emp_total AS (
  SELECT Emp_ID, SUM(sales_amount) total FROM logins GROUP BY Emp_ID
)
SELECT e.*, et.total FROM employees e LEFT JOIN emp_total et ON e.Emp_ID = et.Emp_ID
```

> Hint: 
```sql
WITH RECURSIVE dates AS (
  SELECT '2024-01-01' AS dt
  UNION ALL
  SELECT dt + INTERVAL 1 DAY FROM dates WHERE dt < '2024-01-10'
)
SELECT * FROM dates;
```
