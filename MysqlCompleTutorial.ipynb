## 1. Introduction & Setup

### 1.1 What is MySQL?
MySQL is the world’s most popular open‑source relational database management system (RDBMS). It uses Structured Query Language (SQL) and is widely used in web applications, data warehousing, and enterprise systems.

- **Default engine:** InnoDB (ACID‑compliant, row‑level locking)
- **Default port:** 3306
- **Latest stable:** MySQL 8.4 LTS (2024)

## 5. UPSERT Operations

**UPSERT = UPDATE + INSERT** – insert a row, or update it if it already exists.

### MySQL: `INSERT ... ON DUPLICATE KEY UPDATE`

```sql
-- Basic syntax
INSERT INTO users (email, name, login_count)
VALUES ('john@example.com', 'John Doe', 1)
ON DUPLICATE KEY UPDATE
    name = VALUES(name),
    login_count = login_count + 1,
    last_login = NOW();

-- Modern syntax (MySQL 8.0.19+) using alias
INSERT INTO users (email, name, login_count)
VALUES ('john@example.com', 'John Doe', 1) AS new
ON DUPLICATE KEY UPDATE
    name = new.name,
    login_count = users.login_count + 1,
    last_login = NOW();

-- Multiple rows UPSERT
INSERT INTO inventory (product_id, quantity, location) VALUES
    (101, 50, 'Warehouse A'),
    (102, 30, 'Warehouse A'),
    (103, 75, 'Warehouse B') AS new
ON DUPLICATE KEY UPDATE
    quantity = new.quantity,
    location = new.location;
```

### MySQL: `REPLACE INTO` (Caution!)

```sql
-- Deletes old row and inserts new one (new auto-increment ID!)
REPLACE INTO users (email, name, status)
VALUES ('john@example.com', 'John Doe', 'active');
```

**Warning:** `REPLACE` deletes the existing row and inserts a new one – triggers `DELETE` and `INSERT` triggers, and assigns a new `AUTO_INCREMENT` value. Prefer `ON DUPLICATE KEY UPDATE`.

## 6. DCL – Data Control Language

````python
-- Create users
CREATE USER 'app_user'@'192.168.1.%' IDENTIFIED BY 'StrongP@ssw0rd!';
CREATE USER 'readonly'@'localhost' IDENTIFIED BY 'readonly123';

-- Grant privileges
GRANT SELECT ON mydb.* TO 'readonly'@'localhost';
GRANT CREATE, SELECT ON mydb.* TO 'app_user'@'192.168.1.%';
GRANT INSERT ON mydb.age TO 'app_user'@'192.168.1.%';
GRANT SELECT (id, name) ON mydb.users TO 'app_user'@'192.168.1.%';
GRANT EXECUTE ON PROCEDURE mydb.GetSalary TO 'app_user'@'%';

-- Show grants
SHOW GRANTS FOR 'readonly'@'localhost';

-- Change password
ALTER USER 'app_user'@'192.168.1.%' IDENTIFIED BY 'new_password';

-- Revoke privileges
REVOKE INSERT ON mydb.age FROM 'app_user'@'192.168.1.%';

-- Drop user
DROP USER 'readonly'@'localhost';

-- Role-based permissions (MySQL 8.0+)
CREATE ROLE data_analyst;
GRANT SELECT ON mydb.* TO data_analyst;
CREATE USER 'analyst'@'%' IDENTIFIED BY 'analyst123';
GRANT data_analyst TO 'analyst'@'%';
SET DEFAULT ROLE data_analyst TO 'analyst'@'%';

-- Reload privileges
FLUSH PRIVILEGES;
````

## 7. TCL – Transaction Control Language

````python
-- Basic transaction
START TRANSACTION;
UPDATE accounts SET balance = balance - 500 WHERE id = 1;
UPDATE accounts SET balance = balance + 500 WHERE id = 2;
COMMIT;

-- Rollback
START TRANSACTION;
DELETE FROM employees WHERE emp_id = 5;
ROLLBACK;

-- SAVEPOINT
START TRANSACTION;
INSERT INTO orders (customer_id, amount) VALUES (1, 250.00);
SAVEPOINT after_order;
INSERT INTO payments (order_id, amount) VALUES (LAST_INSERT_ID(), 250.00);
ROLLBACK TO after_order;   -- keep order, undo payment
COMMIT;

-- Isolation levels
SELECT @@transaction_isolation;
SET SESSION TRANSACTION ISOLATION LEVEL READ COMMITTED;
SET SESSION TRANSACTION ISOLATION LEVEL REPEATABLE READ;

-- Locking
START TRANSACTION;
SELECT * FROM accounts WHERE id = 1 FOR UPDATE;
COMMIT;
````

## 8. SQL Clauses – Complete Reference

### Clause Summary Table

| Clause | Purpose | Example |
|--------|---------|---------|
| SELECT | Retrieve data | `SELECT name, age FROM customers;` |
| FROM | Specify table | `FROM employees` |
| WHERE | Filter rows | `WHERE age > 30` |
| ORDER BY | Sort results | `ORDER BY salary DESC` |
| GROUP BY | Group rows | `GROUP BY category` |
| HAVING | Filter groups | `HAVING COUNT(*) > 10` |
| JOIN | Combine tables | `INNER JOIN orders ON customers.id = orders.customer_id` |
| LIMIT | Restrict rows | `LIMIT 10` |
| OFFSET | Skip rows | `OFFSET 5` |
| DISTINCT | Unique values | `SELECT DISTINCT department FROM employees` |
| UNION | Combine results | `SELECT name FROM customers UNION SELECT name FROM suppliers` |
| EXISTS | Check subquery | `WHERE EXISTS (SELECT 1 FROM orders WHERE customers.id = orders.customer_id)` |
| IN | Match list | `WHERE department IN ('HR', 'Finance', 'IT')` |
| BETWEEN | Range filter | `WHERE price BETWEEN 100 AND 500` |
| LIKE | Pattern match | `WHERE name LIKE 'A%'` |

### Execution Order of a SQL Query

```
1. FROM / JOIN
2. WHERE
3. GROUP BY
4. HAVING
5. SELECT
6. DISTINCT
7. ORDER BY
8. LIMIT / OFFSET
```

## 13. Regular Expressions (Regex)

### Basic Syntax

```sql
SELECT column FROM table WHERE column REGEXP 'pattern';
```

### Metacharacters

| Symbol | Meaning | Example |
|--------|---------|---------|
| `.` | Any single character | `'a.b'` matches `acb`, `a2b` |
| `^` | Start of string | `'^A'` starts with A |
| `$` | End of string | `'xyz$'` ends with xyz |
| `*` | Zero or more | `'ab*'` matches `a`, `ab`, `abb` |
| `+` | One or more | `'ab+'` matches `ab`, `abb` |
| `?` | Zero or one | `'colou?r'` matches `color`, `colour` |
| `{n}` | Exactly n | `'a{3}'` matches `aaa` |
| `{n,}` | n or more | `'a{2,}'` matches `aa`, `aaa` |
| `{n,m}` | Between n and m | `'a{2,4}'` matches `aa`, `aaa`, `aaaa` |
| `|` | OR | `'Pro|Max'` matches Pro or Max |
| `[abc]` | Any of a,b,c | `'[abc]'` matches a, b, or c |
| `[^abc]` | Not a,b,c | `'[^abc]'` matches anything else |
| `[a-z]` | Range | `'[a-z]'` lowercase letters |
| `[0-9]` | Digits | `'[0-9]'` any digit |

### Examples

```sql
-- Names starting with 'S'
SELECT * FROM customers WHERE name REGEXP '^S';

-- Names ending with 'n'
SELECT * FROM customers WHERE name REGEXP 'n$';

-- Emails containing numbers
SELECT * FROM users WHERE email REGEXP '[0-9]';

-- Phone numbers with exactly 10 digits
SELECT * FROM contacts WHERE phone REGEXP '^[0-9]{10}$';

-- Products with 'Pro' or 'Max'
SELECT * FROM products WHERE name REGEXP 'Pro|Max';

-- Names containing 'son'
SELECT * FROM customers WHERE name REGEXP 'son';

-- Emails ending with .com
SELECT * FROM users WHERE email REGEXP '\\.com$';

-- Case-insensitive by default, use BINARY for case-sensitive
SELECT * FROM customers WHERE BINARY name REGEXP '^A';
```

### REGEXP_LIKE() – Boolean match

```sql
SELECT * FROM users WHERE REGEXP_LIKE(email, '^[a-z0-9._%+-]+@[a-z0-9.-]+\\.(com|org)$');
```

### REGEXP_REPLACE() – Replace using regex

```sql
-- Remove numbers
SELECT REGEXP_REPLACE('Hello123World', '[0-9]', '');  -- HelloWorld

-- Format phone number
SELECT REGEXP_REPLACE('(987) 654-3210', '[^0-9]', ''); -- 9876543210

-- Mask email username
SELECT REGEXP_REPLACE('john.doe@gmail.com', '^[^@]+', '****'); -- ****@gmail.com

-- Censor profanity
SELECT REGEXP_REPLACE('This is a damn bad word!', 'damn|bad', '****');

-- Extract domain from URL
SELECT REGEXP_REPLACE('https://www.example.com', 'https?://(www\\.)?', ''); -- example.com
```

### REGEXP_SUBSTR() – Extract using regex

```sql
-- Extract first number
SELECT REGEXP_SUBSTR('Order ID: 12345, Ref: 67890', '[0-9]+'); -- 12345

-- Extract second number
SELECT REGEXP_SUBSTR('Order ID: 12345, Ref: 67890', '[0-9]+', 1, 2); -- 67890

-- Extract domain from email
SELECT REGEXP_SUBSTR('john.doe@gmail.com', '@[a-z0-9.-]+'); -- @gmail.com

-- Extract year
SELECT REGEXP_SUBSTR('Date: 2024-02-12', '[0-9]{4}'); -- 2024

-- Extract first hashtag
SELECT REGEXP_SUBSTR('Check this out! #Trending #Viral', '#[A-Za-z0-9]+'); -- #Trending

-- Extract file extension
SELECT REGEXP_SUBSTR('report_final.xlsx', '\\.([a-z]+)$'); -- .xlsx
```


## 15. Subqueries & CTEs

````python
-- Scalar subquery
SELECT name FROM customers WHERE customer_id = (SELECT customer_id FROM orders ORDER BY amount DESC LIMIT 1);

-- IN subquery
SELECT * FROM customers WHERE customer_id IN (SELECT DISTINCT customer_id FROM orders);

-- NOT IN
SELECT * FROM customers WHERE customer_id NOT IN (SELECT DISTINCT customer_id FROM orders);

-- EXISTS
SELECT * FROM customers c WHERE EXISTS (SELECT 1 FROM orders o WHERE o.customer_id = c.customer_id);

-- Subquery in FROM (derived table)
SELECT dept, AVG(salary) FROM (SELECT department AS dept, salary FROM employees WHERE active=1) AS t GROUP BY dept;

-- Correlated subquery
SELECT name, (SELECT COUNT(*) FROM orders WHERE customer_id = c.customer_id) AS order_count FROM customers c;
````

````python
-- CTE
WITH high_value_orders AS (
    SELECT customer_id, SUM(amount) AS total_spent
    FROM orders
    GROUP BY customer_id
    HAVING total_spent > 2000
)
SELECT c.name, h.total_spent FROM customers c JOIN high_value_orders h ON c.customer_id = h.customer_id;

-- Recursive CTE (numbers 1 to 10)
WITH RECURSIVE numbers AS (
    SELECT 1 AS n
    UNION ALL
    SELECT n + 1 FROM numbers WHERE n < 10
)
SELECT * FROM numbers;

-- Recursive CTE for hierarchy
WITH RECURSIVE emp_hierarchy AS (
    SELECT id, name, manager_id, 0 AS level FROM employees WHERE manager_id IS NULL
    UNION ALL
    SELECT e.id, e.name, e.manager_id, h.level + 1
    FROM employees e JOIN emp_hierarchy h ON e.manager_id = h.id
)
SELECT * FROM emp_hierarchy;
````

## 16. Window Functions (MySQL 8.0+)

````python
-- Ranking functions
SELECT name, department, salary,
    ROW_NUMBER() OVER (PARTITION BY department ORDER BY salary DESC) AS row_num,
    RANK() OVER (PARTITION BY department ORDER BY salary DESC) AS rank_num,
    DENSE_RANK() OVER (PARTITION BY department ORDER BY salary DESC) AS dense_rank_num
FROM employees;
````

````python
-- Aggregate window functions
SELECT date, sales,
    SUM(sales) OVER (ORDER BY date) AS running_total,
    AVG(sales) OVER (ORDER BY date ROWS BETWEEN 2 PRECEDING AND CURRENT ROW) AS moving_avg_3,
    sales * 100.0 / SUM(sales) OVER () AS pct_of_total
FROM daily_sales;
````

````python
-- LAG / LEAD
SELECT date, sales,
    LAG(sales, 1) OVER (ORDER BY date) AS previous_day,
    LEAD(sales, 1) OVER (ORDER BY date) AS next_day,
    (sales - LAG(sales) OVER (ORDER BY date)) / LAG(sales) OVER (ORDER BY date) * 100 AS growth_pct
FROM daily_sales;
````

## 19. Triggers & Scheduled Events

````python
-- Audit table
CREATE TABLE salary_audit (
    id INT AUTO_INCREMENT PRIMARY KEY,
    emp_id INT,
    old_salary DECIMAL(10,2),
    new_salary DECIMAL(10,2),
    change_time TIMESTAMP DEFAULT CURRENT_TIMESTAMP
);

-- AFTER UPDATE trigger
DELIMITER //
CREATE TRIGGER salary_update_audit
AFTER UPDATE ON employees
FOR EACH ROW
BEGIN
    IF OLD.salary <> NEW.salary THEN
        INSERT INTO salary_audit (emp_id, old_salary, new_salary) VALUES (NEW.id, OLD.salary, NEW.salary);
    END IF;
END //
DELIMITER ;

-- BEFORE INSERT trigger (auto uppercase)
DELIMITER //
CREATE TRIGGER before_employee_insert
BEFORE INSERT ON employees
FOR EACH ROW
BEGIN
    SET NEW.name = UPPER(NEW.name);
END //
DELIMITER ;

-- Show triggers
SHOW TRIGGERS FROM mydb;

-- Drop trigger
DROP TRIGGER IF EXISTS salary_update_audit;
````

## 21. Partitioning

````python
-- RANGE partitioning by amount
CREATE TABLE orders_partitioned (
    order_id INT NOT NULL,
    customer_id INT NOT NULL,
    amount DECIMAL(10,2) NOT NULL,
    PRIMARY KEY (order_id, amount)
) PARTITION BY RANGE (amount) (
    PARTITION p_small VALUES LESS THAN (100),
    PARTITION p_medium VALUES LESS THAN (500),
    PARTITION p_large VALUES LESS THAN (1000),
    PARTITION p_very_large VALUES LESS THAN MAXVALUE
);

-- HASH partitioning
CREATE TABLE employee_partitioned (
    emp_id INT NOT NULL,
    name VARCHAR(100),
    PRIMARY KEY (emp_id)
) PARTITION BY HASH(emp_id) PARTITIONS 4;

-- LIST COLUMNS partitioning
CREATE TABLE dept_partitioned (
    dept_id INT NOT NULL,
    dept_name VARCHAR(100) NOT NULL,
    PRIMARY KEY (dept_id)
) PARTITION BY LIST COLUMNS(dept_name) (
    PARTITION p_sales VALUES IN ('Sales'),
    PARTITION p_it VALUES IN ('IT'),
    PARTITION p_hr VALUES IN ('HR'),
    PARTITION p_other VALUES IN ('Finance', 'Marketing')
);

-- RANGE COLUMNS partitioning by date
CREATE TABLE sales_partitioned (
    sale_id INT NOT NULL,
    sale_date DATE NOT NULL,
    amount DECIMAL(10,2) NOT NULL,
    PRIMARY KEY (sale_id, sale_date)
) PARTITION BY RANGE COLUMNS(sale_date) (
    PARTITION p2023 VALUES LESS THAN ('2024-01-01'),
    PARTITION p2024 VALUES LESS THAN ('2025-01-01'),
    PARTITION p2025 VALUES LESS THAN ('2026-01-01')
);

-- Manage partitions
ALTER TABLE orders_partitioned ADD PARTITION (PARTITION p_new VALUES LESS THAN (2000));
ALTER TABLE orders_partitioned DROP PARTITION p_small;

-- Show partition info
SELECT TABLE_NAME, PARTITION_NAME, TABLE_ROWS FROM INFORMATION_SCHEMA.PARTITIONS WHERE TABLE_NAME = 'orders_partitioned';
````

## 22. JSON Support

````python
-- Create JSON column
CREATE TABLE products (
    id INT PRIMARY KEY AUTO_INCREMENT,
    name VARCHAR(100),
    attributes JSON
);

-- Insert JSON
INSERT INTO products (name, attributes) VALUES
    ('Laptop', '{"brand": "Dell", "ram": 16, "storage": 512}'),
    ('Phone', JSON_OBJECT('brand', 'Apple', 'storage', 256, 'tags', JSON_ARRAY('mobile', '5g')));

-- Extract values
SELECT name, attributes->>'$.brand' AS brand, attributes->'$.ram' AS ram FROM products;

-- Filter by JSON
SELECT * FROM products WHERE attributes->>'$.brand' = 'Apple';

-- Update JSON
UPDATE products SET attributes = JSON_SET(attributes, '$.ram', 32) WHERE id = 1;
UPDATE products SET attributes = JSON_INSERT(attributes, '$.color', 'silver') WHERE id = 1;
UPDATE products SET attributes = JSON_REMOVE(attributes, '$.color') WHERE id = 1;

-- JSON array functions
SELECT JSON_LENGTH(attributes->>'$.tags') FROM products;
SELECT JSON_CONTAINS(attributes->>'$.tags', '"mobile"') FROM products;

-- Index on JSON (generated column)
ALTER TABLE products ADD COLUMN brand VARCHAR(50) GENERATED ALWAYS AS (attributes->>'$.brand') STORED, ADD INDEX idx_brand (brand);
````

## 23. Python Integration with PyMySQL

````python
import pymysql
import pandas as pd

# Connect
conn = pymysql.connect(
    host='localhost', user='root', password='your_password',
    database='test', cursorclass=pymysql.cursors.DictCursor
)
cursor = conn.cursor()

# Create table
cursor.execute('''CREATE TABLE IF NOT EXISTS employees (
    id INT AUTO_INCREMENT PRIMARY KEY,
    name VARCHAR(100) NOT NULL,
    age INT,
    department VARCHAR(50)
)''')

# Insert single
cursor.execute("INSERT INTO employees (name, age, department) VALUES (%s, %s, %s)", ('Alice', 30, 'HR'))

# Insert multiple
data = [('Bob', 25, 'IT'), ('Charlie', 28, 'Finance')]
cursor.executemany("INSERT INTO employees (name, age, department) VALUES (%s, %s, %s)", data)
conn.commit()

# Query
cursor.execute("SELECT * FROM employees WHERE department = %s", ('IT',))
for row in cursor.fetchall():
    print(row)

# Pandas
df = pd.read_sql("SELECT * FROM employees", conn)
print(df)

# Transaction with savepoint
try:
    cursor.execute("START TRANSACTION")
    cursor.execute("UPDATE accounts SET balance = balance - 500 WHERE id = 1")
    cursor.execute("SAVEPOINT after_debit")
    cursor.execute("UPDATE accounts SET balance = balance + 500 WHERE id = 2")
    conn.commit()
except Exception as e:
    cursor.execute("ROLLBACK TO after_debit")
    conn.commit()
    print("Rolled back", e)

cursor.close()
conn.close()
````

## 27. Security Best Practices
